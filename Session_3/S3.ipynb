# 5. Use LabelEncoder from sklearn to encode the 'player_role' column (e.g., Batsman, Bowler, Allrounder) in your IPL dataset, and display the mapping from original roles to encoded values.<br><br><em><strong>Hint:</strong> Use LabelEncoder's classes_ attribute to see the mapping.

```python
df.columns
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
df['toss_label_encoded'] = le.fit_transform(df['toss_decision'])
```

```python
print("--- Label Encoding Mapping ---")
for index, class_label in enumerate(le.classes_):
    print(f"Original Value: '{class_label}' ===> Encoded Integer: {index}")
```
