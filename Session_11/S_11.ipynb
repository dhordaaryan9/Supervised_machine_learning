# 1. Implement a K-Nearest Neighbors (KNN) classifier in Python to predict whether a Flipkart product review is 'positive' or 'negative' using a small dataset of review texts and their labels.<br><br><em><strong>Hint:</strong> Use scikit-learn's KNeighborsClassifier and convert the text data into numerical features using CountVectorizer.</em>

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
data = {
    'review': [
        'Great product, loved it!', 'Very bad quality, worst choice', 
        'Excellent build and fast delivery', 'Disappointed, not worth the money',
        'Awesome experience, highly recommend', 'Defective item, returning it',
        'Good value for money', 'Terrible customer service and poor item',
        'Amazing performance', 'Waste of money'
    ],
    'sentiment': ['positive', 'negative', 'positive', 'negative', 'positive', 
                  'negative', 'positive', 'negative', 'positive', 'negative']
}
```

```python
df = pd.DataFrame(data)
df.head()
```

```python
X = df['review']
y = df['sentiment']
```

```python
vectorizer = CountVectorizer()
```

```python
X_vectors = vectorizer.fit_transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_vectors, y, test_size=0.3, random_state=42)
```

```python
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)
```

```python
predictions = knn.predict(X_test)
```

```python
print("Accuracy:", accuracy_score(y_test, predictions))
```

# 2. Compare the accuracy of KNN using Euclidean distance vs. Manhattan distance on a dataset of Spotify song features (e.g., tempo, danceability, energy) to classify songs as 'workout' or 'chill' playlists.<br><br><em><strong>Hint:</strong> Use the 'metric' parameter in KNeighborsClassifier to switch between 'euclidean' and 'manhattan'.</em>

```python
# Creating mock Spotify data (tempo, danceability, energy)
# playlist (1 for workout, 0 for chill)
np.random.seed(42)
X_features = np.random.rand(40, 3) * [180, 1, 1]
y_labels = np.random.choice([0, 1], size=40)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_features, y_labels, test_size=0.2, random_state=42)
```

```python
knn_euclidean = KNeighborsClassifier(n_neighbors=3, metric='euclidean')
knn_euclidean.fit(X_train, y_train)
```

```python
pred_euclidean = knn_euclidean.predict(X_test)
```

```python
knn_manhattan = KNeighborsClassifier(n_neighbors=3, metric='manhattan')
knn_manhattan.fit(X_train, y_train)
```

```python
pred_manhattan = knn_manhattan.predict(X_test)
```

```python
print("Euclidean Accuracy:", accuracy_score(y_test, pred_euclidean))
print("Manhattan Accuracy:", accuracy_score(y_test, pred_manhattan))
```

# 4. Build a Multinomial Naive Bayes classifier to categorize WhatsApp chat messages as 'personal', 'group', or 'spam' using example message texts and labels.<br><br><em><strong>Constraint:</strong> Use scikit-learn's MultinomialNB and show the confusion matrix for your results.</em>

```python
data = {
    'message': [
        'Hey, are we still meeting for lunch today?', 'Please send the group project files here',
        'CONGRATULATIONS! You won a cash prize click here', 'Mom says buy milk on your way back',
        'Can everyone check the pinned message in the group?', 'Get cheap loans instantly call now',
        'Let us plan a trip this weekend guys', 'Where are you right now?',
        'Urgent update regarding tomorrows group meeting', 'Earn 50000 rupees sitting at home links below'
    ],
    'category': ['personal', 'group', 'spam', 'personal', 'group', 
                 'spam', 'group', 'personal', 'group', 'spam']
}
```

```python
df = pd.DataFrame(data)
```

```python
X = df['message']
y = df['category']
```

```python
vectorizer = CountVectorizer()
X_vectors = vectorizer.fit_transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_vectors, y, test_size=0.3, random_state=42)
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import confusion_matrix
```

```python
mnb = MultinomialNB()
mnb.fit(X_train, y_train)
```

```python
pred = mnb.predict(X_test)
```

```python
print("Accuracy:", accuracy_score(y_test, pred))
print(confusion_matrix(y_test, pred))
```

Error encountered: ValueError: Found array with dim 3, while dim <= 2 is required by KNeighborsClassifier.  

How I fixed it: Scikit-learn's machine learning models expect a 2D array of shape (n_samples, n_features). The generated image data was 3D (10, 28, 28). I fixed it by flattening the images using .reshape(10, 28*28) to convert each 2D image matrix into a 1D feature array before passing it to the KNN model.

```python
np.random.seed(42)

X_images = np.random.rand(10, 28, 28)
y_labels = ['bat', 'football', 'bat', 'football', 'bat', 'football', 'bat', 'football', 'bat', 'football']

X_flattened = X_images.reshape(10, -1)

knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_flattened, y_labels)

new_image = np.random.rand(1, 28, 28)
new_flattened = new_image.reshape(1, -1)
prediction = knn.predict(new_flattened)

print(f"Prediction for new image: {prediction[0]}")
```
