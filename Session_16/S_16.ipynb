# 1. Given the following confusion matrix from a spam filter (Actual: Spam=70, Not Spam=130; Predicted: Spam=60, Not Spam=140; TP=50, TN=120, FP=10, FN=20), manually calculate the values of accuracy, precision, recall, and F1 score on paper or in a Jupyter notebook.

```python
tp = 50
tn = 120
fp = 10
fn = 20
```

```python
accuracy = (tp + tn) / (tp + tn + fp + fn)
precision = tp / (tp + fp)
recall = tp / (tp + fn)
f1 = 2 * (precision * recall) / (precision + recall)
```

```python
print(f"Accuracy:  {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1 Score:  {f1:.4f}")
```

Calculate and print the precision and recall for this model.<br><br><em><strong>Hint:</strong> Precision = TP / (TP + FP), Recall = TP / (TP + FN)</em>

```python
tp = 60
fn = 30
tn = 120
fp = 10
```

# 4. Given a confusion matrix for a food delivery fraud detection system (TP=40, TN=150, FP=20, FN=10), write Python code to compute the F1 score and explain in one line what the F1 score tells you about the model's performance.<br><br><em><strong>Hint:</strong> F1 = 2 * (precision * recall) / (precision + recall)</em>

```python
tp = 40
tn = 150
fp = 20
fn = 10
```

```python
precision = tp / (tp + fp)
recall = tp / (tp + fn)
```

```python
f1_score = 2 * (precision * recall) / (precision + recall)
```

```python
print("Calculated F1 Score:", round(f1_score, 4))
```

**What the F1 Score tells you**:
The F1 score gives you a single balanced performance metric that proves the model effectively flags real delivery frauds without frustrating innocent customers with excessive false alarms.
