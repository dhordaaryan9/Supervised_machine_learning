# 1. Load the Iris dataset using scikit-learn and use cross_val_score() with a KNeighborsClassifier to compute the average accuracy using 5-fold cross-validation.


```python
from sklearn.datasets import load_iris
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score
```

```python
iris = load_iris()
X, y = iris.data, iris.target
```

```python
knn = KNeighborsClassifier(n_neighbors=3)
```

```python
scores = cross_val_score(knn, X, y, cv=5)
```

```python
print("Scores for each fold:", scores)
print("Average Accuracy   :", scores.mean())
```

# 2. For a dataset of your choice (e.g., Titanic or any from scikit-learn), perform 10-fold cross-validation using cross_val_score() with a DecisionTreeClassifier and print the accuracy for each fold.

```python
from sklearn.datasets import load_wine
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score
```

```python
wine = load_wine()
X, y = wine.data, wine.target
```

```python
tree = DecisionTreeClassifier(random_state=42)
```

```python
scores_10fold = cross_val_score(tree, X, y, cv=10)
```

```python
print("Accuracy for each of the 10 folds:")
for idx, score in enumerate(scores_10fold):
    print(f"Fold {idx+1}: {round(score, 4)}")
print(f"Mean Accuracy: {round(scores_10fold.mean(), 4)}")
```

# 3. Use StratifiedKFold from scikit-learn to split the Iris dataset and compare the distribution of target classes in each fold to show how stratified sampling works.<br><br><em><strong>Hint:</strong> Print the class counts for each fold to verify stratification.</em>

```python
from sklearn.model_selection import StratifiedKFold
```

```python
iris = load_iris()
X, y = iris.data, iris.target
```

```python
skf = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
```

```python
for fold, (train_index, test_index) in enumerate(skf.split(X, y)):
    y_train_fold = y[train_index]
    y_test_fold = y[test_index]
    
    classes, counts = np.unique(y_test_fold, return_counts=True)
    print(f"Fold {fold+1} Test Set Class Distribution: {dict(zip(classes, counts))}")
```

As seen in the output, StratifiedKFold ensures that each fold contains the exact same percentage of each target class as the complete original dataset. Instead of randomly splitting lines—which might leave a fold completely missing a certain flower category—stratification keeps the proportion balanced across all chunks so the model trains and tests fairly on every group.

# 4. Experiment with different values of 'cv' (3, 5, 10) in cross_val_score() for a RandomForestClassifier on the Wine dataset and report which value gives the most stable (least variable) accuracy scores.<br><br><em><strong>Hint:</strong> Look at the standard deviation of the scores for each cv value.</em>

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
wine = load_wine()
X, y = wine.data, wine.target
```

```python
rf = RandomForestClassifier(n_estimators=50, random_state=42)
cv_values = [3, 5, 10]
```

```python
for cv in cv_values:
    scores = cross_val_score(rf, X, y, cv=cv)
    print(f"CV = {cv:<2} | Mean Acc: {round(scores.mean(), 4)} | Std Dev (Variance): {round(scores.std(), 4)}")
```

By comparing the standard deviations across the runs, the CV = 5 (or 10, depending on your random split numbers) value showed the lowest standard deviation. A lower standard deviation means that the accuracy scores fluctuated the least between individual iterations, proving that this specific fold constraint offers the most stable and reliable evaluation framework for our data matrix.
