# 1. Load the Iris dataset using scikit-learn, train a DecisionTreeClassifier, and plot the resulting decision tree using plot_tree().


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
iris = load_iris()
```

```python
X = iris.data
y = iris.target
```

```python
tree_model = DecisionTreeClassifier(random_state=42)
tree_model.fit(X, y)
```

```python
plt.figure(figsize=(12, 8))
plot_tree(tree_model, feature_names=iris.feature_names, class_names=iris.target_names, filled=True)
plt.title("Full Depth Decision Tree on Iris Dataset")
plt.show()
```

# 2. Change the splitting criterion of your DecisionTreeClassifier from 'gini' to 'entropy' and compare the accuracy scores for both on the same dataset.<br><br><em><strong>Hint:</strong> Use the 'criterion' parameter when initializing the classifier.</em>

```python
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.3, random_state=42)
```

```python
model_gini = DecisionTreeClassifier(criterion='gini', random_state=42)
model_gini.fit(X_train, y_train)
```

```python
pred_gini = model_gini.predict(X_test)
```

```python
model_entropy = DecisionTreeClassifier(criterion='entropy', random_state=42)
model_entropy.fit(X_train, y_train)
```

```python
pred_entropy = model_entropy.predict(X_test)
```

```python
print("Gini Accuracy    :", accuracy_score(y_test, pred_gini))
print("Entropy Accuracy :", accuracy_score(y_test, pred_entropy))
```

**Comparison**:
Both gini and entropy often produce very similar or identical accuracy on standard datasets like Iris. Gini measures the probability of misclassifying a random element, while Entropy measures information gain. Gini is the default because it is slightly faster to compute (no logarithms required).

# 3. Limit the maximum depth of your decision tree to 2 and visualize the new tree structure. Observe and briefly describe how the tree's decisions have changed compared to the full-depth tree.

```python
tree_limited = DecisionTreeClassifier(max_depth=2, random_state=42)
tree_limited.fit(iris.data, iris.target)
```

```python
plt.figure(figsize=(10, 6))
plot_tree(tree_limited, feature_names=iris.feature_names, class_names=iris.target_names, filled=True)
plt.title("Decision Tree Limited to Max Depth = 2")
plt.show()
```

By setting max_depth=2, the tree structure becomes much smaller and simpler compared to the full-depth tree. The full-depth tree continues splitting down until every single training sample is perfectly isolated into pure leaf nodes, which can lead to overfitting. The limited tree stops growing after just two structural layers of decisions. While it might misclassify a couple of tricky overlapping edge cases, it focuses purely on the most dominant data variations, making it much more interpretable and stable.

# 4. Use the feature_importances_ attribute of your trained DecisionTreeClassifier to display the importance score for each feature in the Iris dataset. Sort and print the features from most to least important.

```python
model = DecisionTreeClassifier(random_state=42)
model.fit(iris.data, iris.target)
```

```python
importances = model.feature_importances_
```

```python
df_importance = pd.DataFrame({
    'Feature': iris.feature_names,
    'Importance': importances
})
```

```python
df_sorted = df_importance.sort_values(by='Importance', ascending=False)
```

```python
print(df_sorted.to_string(index=False))
```
