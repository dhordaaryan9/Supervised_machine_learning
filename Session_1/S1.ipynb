# 2. Given the following scenarios, classify each as supervised, unsupervised, or reinforcement learning: 1) Netflix recommending movies based on your watch history, 2) Spotify grouping similar songs into playlists, 3) A self-driving car learning to park by trial and error. Write your answers with a one-line explanation for each choice.
---
1) Netflix recommending movies: Supervised Learning.   
***Explanation***: The model predicts future preferences by using your labeled past watch history and ratings as a guide.

2) Spotify grouping similar songs: Unsupervised Learning.     
***Explanation***: The system relies on clustering algorithms to uncover hidden patterns and group songs without using pre-existing labels.

3) A self-driving car learning to park: Reinforcement Learning.    
***Explanation***: An autonomous agent learns the optimal sequence of actions by interacting with the environment through a system of trial, error, rewards, and penalties

# 4. Imagine you are building a spam filter for WhatsApp messages. Write a short paragraph explaining how traditional programming would approach this problem versus how machine learning would approach it.<br><br><em><strong>Hint:</strong> Focus on rules vs learning from data.

**Traditional Programming:** A software developer must manually write strict rules (like if message contains "free cash"). The system only blocks exact matches and fails if spammers change their wording.     

**Machine Learning:** You feed the computer thousands of past messages labeled as "spam" or "not spam." The system learns the patterns on its own, allowing it to automatically catch new, sneaky variations without human updates.
